# 05. Візуалізація уваги BERT для шаблонів логів

Цей ноутбук виконує якісний аналіз: візуалізує матриці уваги BERT для обраних шаблонів ZooKeeper, щоб показати, на які токени модель спирається під час побудови подання `[CLS]`. Це доповнює кількісні метрики з ноутбука 04 та допомагає інтерпретувати, чому певні шаблони потрапляють до одного семантичного кластера. Графіки зберігаються поряд із ноутбуком та використовуються в розділі експериментів дипломної роботи.
